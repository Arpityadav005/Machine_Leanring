```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')
```

```python
df = pd.read_csv("amazon.csv")
df
```

```python
df.info()
```

```python
#1. What is the average rating for each product category?
df.rating.value_counts()
#since data type is int we can convert it into int
df.rating = pd.to_numeric(df.rating,errors = 'coerce')
df.rating = df.rating.astype(float)
df.groupby("category")['rating'].mean().round(2)
```

```python
#2. What are the top rating_count products by category?
df['rating_count'] = df['rating_count'].str.replace(",", "").astype(float)
```

```python
# 1. Find the index of top product in each category
idx = df.groupby('category')['rating_count'].idxmax()

# 2. Fetch the full rows using loc
top_products = df.loc[idx, ['category', 'product_name', 'rating_count']]

# 3. Optional: sort overall by rating_count descending
top_products = top_products.sort_values(by='rating_count', ascending=False).reset_index(drop=True)

top_products
```

```python
#3. What is the distribution of discounted prices vs. actual prices?
df['discounted_price'] = df['discounted_price'].str.replace(r"[₹,]", "", regex=True).astype(float)
df['actual_price'] = df['actual_price'].str.replace(r"[₹,]", "", regex=True).astype(float)
df.info()
```

```python
sns.violinplot(data=df[['actual_price','discounted_price']])
# plt.xticks([0,1], ['Actual Price', 'Discounted Price'])
plt.ylim(0, 145000)  # adjust according to your typical price range
plt.title("Distribution of Actual vs Discounted Prices (Focus on Majority)")
plt.show()
```

```python
#4. How does the average discount percentage vary across categories?
df.discount_percentage = df.discount_percentage.str.strip("%")
```

```python
df.discount_percentage = df.discount_percentage.astype(float)
```

```python
df1 = df.groupby('category')['discount_percentage'].mean().reset_index(name = 'avg_discount')
df1 = df1.sort_values(by='avg_discount', ascending=False).reset_index(drop=True)
df1.columns
```

```python
df1['top_category'] = df1['category'].apply(lambda x: x.split('|')[0])
category_discount = df1.groupby('top_category')['avg_discount'].mean().reset_index()
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(12,6))
sns.barplot(x='top_category', y='avg_discount', data=category_discount)
plt.xticks(rotation=90, ha='right')
plt.title('Average Discount Percentage by Top-Level Category')
plt.ylabel('Average Discount (%)')
plt.xlabel('Category')
plt.show()
```
