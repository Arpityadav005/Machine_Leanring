```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
#reading the data..
#here we are loading the data from skiti-learn
from sklearn.datasets import load_diabetes
```

```python
diabetes = load_diabetes()
```

```python
diabetes
```

```python
#problem statement : To predict disease progession one year after baseline provided 10 features
#in one year how the sugar level has changed
```

```python
print(diabetes.DESCR)
```

```python
diabetes.data
```

```python
diabetes.target
```

```python
diabetes.feature_names
```

```python
data = pd.DataFrame(diabetes.data,columns= diabetes.feature_names)
```

```python
data.head()
```

```python
data['target'] = diabetes.target
```

```python
data
```

```python
#EDA,Data cleaning, data preparation, feature enigneering
```

```python
X = data[['bmi']]  # one independent variable
y = data['target'] # dependent variable 
```

```python
X
```

```python
y
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(X,y,test_size = 0.2, random_state = 1)
#random_state when fixed help us to reproduce the same result again and again
```

```python
X_train
```

```python
X_train.shape
```

```python
X_test.shape
```

```python
y_train.shape
```

```python
y_test.shape
```

```python
#scaling(optional)
```

```python
#model training
from sklearn.linear_model import LinearRegression
```

```python
model = LinearRegression()
```

```python
model
```

```python
X_train
```

```python
y_train
```

```python
#training of model
model.fit(X_train, y_train)
# at the backend, it will use gradient descent to give optional coefficient
```

```python
model.coef_ # attribute of the model
```

```python
model.intercept_
```

```python
y_pred = model.predict(X_test)
```

```python
#visualise the result
plt.scatter(X_test, y_test, color = 'black', label = "Actual Data")
plt.plot(X_test,y_pred,color = 'Blue', linewidth = 3,label = "Predicted linear regression line")
plt.xlabel("BMI")
plt.ylabel("One Year Progession target")
plt.title("Linear regression on diabetes data")
plt.legend()
plt.show()
```

```python
#model evaluation
from sklearn.metrics import mean_squared_error
mean_squared_error(y_test,y_pred)
```
