```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

import warnings
warnings.filterwarnings('ignore')
```

```python
from sklearn.datasets import load_iris
```

```python
data = load_iris()
data
```

```python
#logistic regression till  now we have studied, with two values of target variable >> binary classification
```

```python
df=pd.DataFrame(data.data, columns = data.feature_names)
df['target'] = data.target
```

```python
df
```

```python
# ML Pipeline-> Data ingestion,eda/feature engineering,data preparation, seperate in x and y
#>train_test_split,scaling(optional),modeling,evaluation
```

```python
df.head()
```

```python
df.sample(3)
```

```python
df.target.unique()
```

```python
#for binary classification>> ony two call needed, so we will filter out class 2
df=df[df['target'] != 2]
df
```

```python
df.target.unique()
```

```python
X = df.iloc[:,:-1]
y = df.iloc[:,-1]
```

```python
#train test split
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test=train_test_split(X,y,test_size=0.20,random_state=1)
```

```python
X_train.shape , X_test.shape
```

```python
#model training
from sklearn.linear_model import LogisticRegression
classifier = LogisticRegression()
classifier
```

```python
classifier.fit(X_train, y_train)
```

```python
y_pred = classifier.predict(X_test)
y_pred
```

```python
classifier.predict_proba(X_test)
```

```python
#evaluation matrics
from sklearn.metrics import confusion_matrix, accuracy_score, classification_report
```

```python
confusion_matrix(y_test,y_pred)
```

```python
accuracy_score(y_test,y_pred)
```

```python
print(classification_report(y_test,y_pred))
```

```python
#first two rows, 0 and 1 >> precision recall, f1 score and np of data points with respect to respective class
#macro avg -> for each class take simple avg >> avg precision of class 0 and class 1 will be macro average
#similary for recall, f1 score
#support will be total no of datapoints

#weighted avg is used in case of class imbalance, it gives importance to performance of dominant class
```

```python
from sklearn.metrics import roc_curve, auc

#get predicted probabilites of class 1 
classifier.predict_proba(X_test)[:,1] # probability for one class
```

```python
y_pred_proba = classifier.predict_proba(X_test)[:,1]
```

```python
y_pred_proba
```

```python
#Reciever operating characterstic curve
#input : y_test, y_pred_proba. It will return TPR, FPR with different cutoff of probabilities

fpr, tpr, thresholds = roc_curve(y_test,y_pred_proba)
```

```python
thresholds
```

```python
tpr
```

```python
fpr
```

```python
roc_auc = auc(fpr,tpr)
roc_auc
```

```python
# Plot ROC-AUC curve
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 6))  # Create a new figure with a specified size
plt.plot(fpr, tpr, color='darkorange', linewidth=2, label='ROC curve (area = %0.2f)' % roc_auc)  # Plot ROC curve
plt.plot([0, 1], [0, 1], color='navy', linewidth=2, linestyle='--')  # Plot the diagonal line representing random guessing
plt.xlim([0.0, 1.0])  # Set x-axis limits
plt.ylim([0.0, 1.05])  # Set y-axis limits
plt.xlabel('False Positive Rate')  # Set x-axis label
plt.ylabel('True Positive Rate')  # Set y-axis label
plt.title('Receiver Operating Characteristic (ROC) Curve')  # Set plot title
plt.legend(loc="lower right")  # Add legend to the plot
plt.show()  # Show the plot
```
