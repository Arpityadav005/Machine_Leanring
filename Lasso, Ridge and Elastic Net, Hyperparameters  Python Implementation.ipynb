```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

import warnings
warnings.filterwarnings('ignore')
```

```python
df = sns.load_dataset('mpg')
```

```python
df
```

```python
df.drop('name',axis=1,inplace=True)
```

```python
df.head()
```

```python
df.info()
```

```python
df.isna().sum()
```

```python
#since we have not done outliar treatement , then better idea will be to replace the missing value with midean
```

```python
df.horsepower.median()
```

```python
df.horsepower = df.horsepower.fillna(df.horsepower.median())
```

```python
df.isna().sum()
```

```python
df.origin.value_counts()
```

```python
#data encoding -- to convert categorical data to numerical
```

```python
df['origin'] = df['origin'].map({'usa':1,'japan':2,'europe':3})
```

```python
df['origin'] = df['origin'].astype(int)
```

```python
df.info()
```

```python
#seperate into X and y
```

```python
X = df.drop('mpg',axis=1)
y = df['mpg']
```

```python
X
```

```python
y
```

```python
#train test split
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X,y,test_size = 0.3,random_state=1)
```

```python
X_train.shape
```

```python
#simple Linear regression model
from sklearn.linear_model import LinearRegression
regression_model = LinearRegression()
```

```python
regression_model.fit(X_train,y_train)
```

```python
regression_model.coef_
```

```python
regression_model.intercept_
```

```python
for i, col_name in enumerate(X.columns):
    print(f"The coefficient for {col_name} is {regression_model.coef_[i]}")
```

```python
#Observations:
#coefficent are relatively smaller, if one indpendent variable changes
#there will be not much difference in prediction
#This is sometime called as smoother model

#These feature might not be contributing in model training
```

```python
from sklearn.metrics import r2_score

y_pred_linear = regression_model.predict(X_test)
r2_linear = r2_score(y_test,y_pred_linear)
print(f'The square of linear regression {r2_linear}')
```

```python
#regularised model
#reidge regression
```

```python
from sklearn.linear_model import Ridge
ridge_regression_model = Ridge(alpha=0.1)
ridge_regression_model

#in practical implementation lambda is alpha
```

```python
ridge_regression_model.fit(X_train,y_train)
```

```python
for i, col_name in enumerate(X.columns):
    print(f"The coefficient for {col_name} is {ridge_regression_model.coef_[i]}")
```

```python
#for ridge regression evaluation
y_pred_ridge = ridge_regression_model.predict(X_test)
r2_ridge = r2_score(y_test,y_pred_ridge)
print(f"The square of linear regression {r2_ridge}")
```

```python
#we dont see much variation in coeff of ridge regression as compared to linear regression
from sklearn.linear_model import Lasso
```

```python
lasso_regression_model.fit(X_train,y_train)
```

```python
for i, col_name in enumerate(X.columns):
    print(f"The coefficient for {col_name} is {lasso_regression_model.coef_[i]}")
```

```python
#Three feature coefficinet is 0, lasso helps in feature selection
```

```python
#CV with hyperparameter tuning
```

```python
from sklearn.model_selection import GridSearchCV, RandomizedSearchCV
from sklearn.linear_model import Lasso, Ridge
```

```python
#gridsearchcv for lasso
#Define the model
lasso = Lasso()
lasso
```

```python
#GridSearchCV is a tool in scikit-learn used to automatically 
#find the best combination of hyperparameters for a model.
```

```python
#grid search cv >> Dictionary with parameters names (str) as keys and lists of parameter settings to try as values

param_grid = {'alpha' : [0.001,0.01,0.1,1,10,100]}
param_grid
```

```python
grid_search = GridSearchCV(estimator=lasso,param_grid=param_grid,cv=5,scoring = 'r2', verbose=2)
```

```python
grid_search
```

```python
grid_search.fit(X_train,y_train)
```

```python
grid_search.best_params_['alpha']
```

```python
grid_search.best_score_
```

```python
grid_search.best_estimator_
#Estimeter that was chosen by the search, i.e estimator which give highest(or smallest loss if specified) on the left out data.
```

```python
y_pred = grid_search.best_estimator_.predict(X_test)
r2_score(y_test,y_pred)
```

```python
#Randomised seach cv for lasso
lasso = Lasso()
param_distribution = {'alpha': [0.001,0.01,0.1,1,10,100]}
random_search_lasso = RandomizedSearchCV(estimator=lasso, param_distributions=param_distribution, cv = 5, scoring = 'r2',verbose = 2 )
random_search_lasso
```

```python
random_search_lasso.fit(X_train,y_train)

#due to n_iter paramets in each cross validation random 3 parameter are selected
```

```python
 random_search_lasso.best_estimator_
```

```python
 random_search_lasso.best_score_
```

```python
y_pred = random_search_lasso.best_estimator_.predict(X_test)
r2_score(y_test,y_pred)
```

```python
#Ridge Grid Search CV
ridge = Ridge()
param_grid = {'alpha' : [0.001,0.01,0.1,1,10,100]}
ridge = GridSearchCV(estimator=ridge, param_grid=param_grid, cv=5, scoring ='r2',verbose =2)
ridge.fit(X_train,y_train)
ridge.best_params_
```

```python
ridge.best_estimator_
```

```python
y_pred = ridge.best_estimator_.predict(X_test)
r2_score(y_test,y_pred)
```

```python
#Ridge Randomiszed seach CV
ridge = Ridge()
param_distributions = {'alpha' : [0.001,0.01,0.1,1,10,100]}
ridge = RandomizedSearchCV(estimator=ridge, param_distributions=param_distributions, cv=5, scoring ='r2',n_iter=2,verbose =2)
ridge.fit(X_train,y_train)
print(ridge.best_params_)
print(ridge.best_estimator_)
y_pred = ridge.best_estimator_.predict(X_test)
print(r2_score(y_test,y_pred))
```
