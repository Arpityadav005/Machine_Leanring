```python
df=df.dropna()
```

```python
df["Day"] = df['Date_of_Journey'].str.split("/").str[0]
df["Month"] = df['Date_of_Journey'].str.split("/").str[1]
df["Year"] = df['Date_of_Journey'].str.split("/").str[2]
```

```python
df.dtypes
```

```python
df["Day"] = df["Day"].astype(int)
df["Month"] = df["Month"].astype(int)
df["Year"] = df["Year"].astype(int)
```

```python
df.drop("Date_of_Journey",axis =1,inplace = True)
```

```python
df.info()
```

```python
df.Arrival_Time.value_counts()
```

```python
df['Arrival_Time'] = df['Arrival_Time'].apply(lambda x:x.split(' ')[0])
#we want 0th element only
```

```python
df["Arrival_hour"] = df["Arrival_Time"].str.split(':').str[0]
df["Arrival_mins"] = df["Arrival_Time"].str.split(':').str[1]
```

```python
df.info()
```

```python
df["Arrival_hour"]=df["Arrival_hour"].astype(int)
df["Arrival_mins"]=df["Arrival_mins"].astype(int)
df.info()
```

```python
df.drop('Arrival_Time',axis=1,inplace = True)
```

```python
df['Dep_hour'] = df['Dep_Time'].apply(lambda x: x.split(":")[0])
df['Dep_mins'] = df['Dep_Time'].apply(lambda x: x.split(":")[1])
df['Dep_hour'] = df['Dep_hour'].astype(int)
df['Dep_mins'] = df['Dep_mins'].astype(int)
df.drop('Dep_Time',axis = 1,inplace = True)
```

```python
df.info()
```

```python
df.Total_Stops
```

```python
df.Total_Stops.value_counts()
```

```python
df['Total_Stops'] = df['Total_Stops'].map({'non-stop':0,'1 stop':1,'2 stops':2,'3 stops':3,'4 stops':4})
```

```python
df.info()
```

```python
#Route can be used in two ways
#to know source and destination>> which is already in the data
#as number of stops >> total stops is aleady there
#so route is a repeatative information
```

```python
df.drop('Route',axis=1,inplace = True)
```

```python
df.info()
```

```python
df.Duration
```

```python
df['duration_hour'] = df.Duration.str.split(' ').str[0].str.split('h').str[0]
```

```python
df['duration_mins'] = df.Duration.str.split(' ').str[1].str.split('m').str[0]
```

```python
df.drop('Duration',axis=1,inplace=True)
```

```python
df.dtypes
```

```python
df.duration_hour.value_counts()
```

```python
df["duration_hour"].replace('5m','5',inplace = True)
```

```python
df["duration_hour"] = df.duration_hour.astype(int)
```

```python
df['duration_mins']=df['duration_mins'].astype(int)
df['duration_hour']=df['duration_hour'].astype(int)
```
